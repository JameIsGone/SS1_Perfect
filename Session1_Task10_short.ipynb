# Session1_Task10_short — Churn Analysis

```python
import pandas as pd, warnings; warnings.filterwarnings('ignore')

c       = pd.read_csv('customers_cleaned.csv')
cltv_df = pd.read_csv('Session1_CLTV.csv')

# normalize churned: True/Yes/1/Y → True, อื่นๆ → False
c['is_churned'] = c['churned'].astype(str).str.lower().str.strip().isin({'true','yes','1','y'})

# churn rate %
churn_rate = round(c['is_churned'].sum() / len(c) * 100, 2)

# avg CLTV แยก churned vs active
avg = c[['customer_id','is_churned']].merge(cltv_df, on='customer_id', how='inner').groupby('is_churned')['cltv'].mean().round(2)

pd.DataFrame([{'churn_rate': churn_rate,
               'avg_cltv_churned': float(avg.get(True,  0)),
               'avg_cltv_active' : float(avg.get(False, 0))
}]).to_csv('Session1_Churn_Analysis_short.csv', index=False)

print(f'Churn rate: {churn_rate}%')
print('✅ Saved Session1_Churn_Analysis_short_short.csv')
# จุดสังเกต: 1 แถว 3 คอลัมน์, churn_rate 0-100, avg_cltv_active > avg_cltv_churned
```
