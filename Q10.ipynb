```python
import pandas as pd, warnings; warnings.filterwarnings('ignore')
c = pd.read_csv('customers_cleaned.csv')
cltv = pd.read_csv('Session1_CLTVVTest.csv')

churned_true = {'true','yes','1','y'}
c['is_churned_true'] = c['churned'].astype(str).str.lower().str.strip().isin(churned_true)
print(c['is_churned_true'].value_counts())
```

```python
churned_rate = round(c['is_churned_true'].sum() / len(c) * 100, 2)
print(f'{churned_rate} %')
```

```python
merged = c[['customer_id','is_churned_true']].merge(cltv, on='customer_id', how='inner')
avg_cltv = merged.groupby('is_churned_true')['cltv'].mean().round(2)

avg_cltv_churned = float(avg_cltv.get(True, 0))
avg_cltv_active = float(avg_cltv.get(False, 0))

print(avg_cltv_churned)
print(avg_cltv_active)
```

```python
result = pd.DataFrame([{
    'churn_rate' : churned_rate,
    'avg_cltv_churned' : avg_cltv_churned,
    'avg_cltv_active' : avg_cltv_active
    
}])

display(result)
```

```python
result.to_csv('Session1_Churn_AnalysisVTest.csv', index=False)
print('Success')
```
