```python
import pandas as pd, warnings; warnings.filterwarnings('ignore')
s = pd.read_csv('sales_transactions_cleaned.csv')
s['revenue'] =  (s['quantity']*s['price'])- pd.to_numeric(s['discount_amount'],errors='coerce').fillna(0)
s['date'] = pd.to_datetime(s['date'])

cust = s.groupby('customer_id').agg(total_revenue = ('revenue', 'sum'), total_tx = ('transaction_id','nunique')).reset_index()
cust['average_purchase_value'] = cust['total_revenue'] / cust['total_tx']
display(cust)
```

```python
total_month = s['date'].dt.to_period('M').nunique()
cust['purchase_frequency'] = cust['total_tx'] / total_month

cust['cltv'] = (cust['average_purchase_value'] * cust['purchase_frequency'] * 36).round(2)

result = cust[['customer_id','cltv']].copy()
result['customer_id'] = result['customer_id'].astype(int)
display(result)
```

```python
result.to_csv('Session1_CLTVVTest.csv', index=False)
print('Success')
```
