# Session1_Task6_short — Time Series Forecasting (ARIMA)

```python
import pandas as pd, numpy as np, warnings; warnings.filterwarnings('ignore')
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_absolute_error

s = pd.read_csv('sales_transactions_cleaned.csv')
s['date']    = pd.to_datetime(s['date'])
s['revenue'] = (s['quantity']*s['price']) - pd.to_numeric(s['discount_amount'],errors='coerce').fillna(0)

# daily revenue
daily = (s.groupby(s['date'].dt.date)['revenue'].sum()
          .rename_axis('Date').rename('Sales')
          .pipe(lambda x: x.set_axis(pd.to_datetime(x.index))).sort_index())

# train/test split 80/20 → คำนวณ MAE
split        = int(len(daily)*0.8)
train, test  = daily.iloc[:split], daily.iloc[split:]
res_test     = ARIMA(train, order=(1,1,1)).fit()                         # fit บน train
mae          = mean_absolute_error(test, res_test.get_forecast(steps=len(test)).predicted_mean)
print(f'MAE: {mae:.2f}')

# forecast 30 วันข้างหน้า (ใช้ข้อมูลทั้งหมด)
res_full     = ARIMA(daily, order=(1,1,1)).fit()                         # fit บน full data
fc           = res_full.get_forecast(steps=30).predicted_mean            # ค่าทำนาย 30 วัน
future_dates = pd.date_range(daily.index.max()+pd.Timedelta(days=1), periods=30, freq='D')

pd.DataFrame({'Date': future_dates.strftime('%Y-%m-%d'), 'Predicted_Sales': fc.values.round(2)
}).to_csv('Session1_SalesForecast_short.csv', index=False)
print('✅ Saved Session1_SalesForecast_short.csv')
# จุดสังเกต: ไฟล์มี 30 แถว, Predicted_Sales ไม่ติดลบ, MAE สมเหตุสมผล
```
