```python
import pandas as pd, numpy as np, warnings; warnings.filterwarnings('ignore')

from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_absolute_error
```

```python
s = pd.read_csv('sales_transactions_cleaned.csv')
s['date'] = pd.to_datetime(s['date'])
s['revenue'] =  (s['quantity']*s['price'])- pd.to_numeric(s['discount_amount'],errors='coerce').fillna(0)

daily = (s.groupby(s['date'].dt.date)['revenue'].sum().reset_index().rename(columns={'date':'Date','revenue':'Sales'}))
daily['Date'] = pd.to_datetime(daily['Date'])
daily = daily.set_index('Date').sort_index()

print(daily.shape)
print(daily.index.min(),daily.index.max())
display(daily.head())
```

```python
split = int(len(daily) * 0.8)
train = daily.iloc[:split]
test = daily.iloc[split:]

print(len(train))
print(len(test))

model = ARIMA(train['Sales'], order=(1,1,1))
result = model.fit()

testpred = result.get_forecast(steps=len(test)).predicted_mean

mae = mean_absolute_error(test['Sales'], testpred)
print(f'MAE: {mae:.2f}')
```

```python
full_model = ARIMA(daily['Sales'], order=(1,1,1))
full_result = full_model.fit()
forecast30 = full_result.get_forecast(steps=30).predicted_mean

lastdate = daily.index.max()
futuredate = pd.date_range(start=lastdate + pd.Timedelta(days=1), periods=30, freq='D')

forecastdf = pd.DataFrame({'Date' : futuredate.strftime('%Y-%m-%d'), 'Predicted_Sales' : forecast30.values.round(2)})

display(forecastdf)

forecastdf.to_csv('Session1_SalesForecastVtest.csv', index=False)
```
