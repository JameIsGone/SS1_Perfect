```python
import pandas as pd
import matplotlib.pyplot as plt 
import matplotlib.ticker as mticker 
from matplotlib.backends.backend_pdf import PdfPages

c = pd.read_csv('customers_cleaned.csv')
```

```python
c['age_group'] = pd.cut(c['age'],bins=[17, 24, 34, 44, 1000], labels=['18-24', '25-34', '35-44', '45+'])
agecount = c['age_group'].value_counts().sort_index()
print(agecount)
```

```python
gendercount = c[c['gender'].isin(['M','F'])]['gender'].value_counts()
genderpct = (gendercount / gendercount.sum() * 100).round(2)
genderdf = pd.DataFrame({'Gender': genderpct.index, 'Percentage (%)': genderpct.values})
display(genderdf)
```

```python
c['membership_status'] = c['membership_status'].str.title()

tier = c[c['membership_status'].isin({'Basic','Silver','Gold'})]

avgspend = (tier.groupby('membership_status')['total_spending'].mean().round(2).reindex(['Basic','Silver','Gold']))
tierdf = pd.DataFrame({'Tier': avgspend.index, 'Avg Spending ($)' : avgspend.values})
display(tierdf)
```
