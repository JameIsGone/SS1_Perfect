# Session1_Task7_short — Customer Segmentation & Recommendation

```python
import pandas as pd, numpy as np, warnings; warnings.filterwarnings('ignore')
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from itertools import combinations
from collections import Counter

s = pd.read_csv('sales_transactions_cleaned.csv')
c = pd.read_csv('customers_cleaned.csv')
s['revenue'] = (s['quantity']*s['price']) - pd.to_numeric(s['discount_amount'],errors='coerce').fillna(0)

# Feature engineering: total_purchases, avg_purchase_value ต่อลูกค้า
feat = s.groupby('customer_id').agg(
    total_purchases=('transaction_id','nunique'),
    avg_purchase_value=('revenue','mean')
).reset_index()

# KMeans clustering: StandardScaler ปรับ scale ก่อน, n_clusters=3
X = StandardScaler().fit_transform(feat[['total_purchases','avg_purchase_value']])
feat['cluster_label'] = KMeans(n_clusters=3, random_state=42, n_init=10).fit_predict(X) + 1

# Product affinity: นับคู่สินค้าที่ซื้อพร้อมกัน
pairs = Counter()
for prods in s.groupby('transaction_id')['product_id'].apply(list):
    if len(prods)>=2: [pairs.__setitem__(p, pairs.get(p,0)+1) for p in combinations(sorted(prods),2)]

affinity = {}
for pid in s['product_id'].dropna().unique():
    rel = {b:c for (a,b),c in pairs.items() if a==pid}
    rel.update({a:c for (a,b),c in pairs.items() if b==pid})
    affinity[pid] = sorted(rel, key=rel.get, reverse=True)[:3]

# Recommendation: top products ในกลุ่มเดียวกัน ที่ยังไม่เคยซื้อ
purchased   = s.groupby('customer_id')['product_id'].apply(set).to_dict()
s_cl        = s.merge(feat[['customer_id','cluster_label']], on='customer_id')
cluster_top = (s_cl.groupby(['cluster_label','product_id'])['quantity'].sum()
               .reset_index().sort_values(['cluster_label','quantity'],ascending=[True,False]))

rows = []
for _, row in feat[['customer_id','cluster_label']].iterrows():
    cid, cl = row['customer_id'], row['cluster_label']
    recs = [int(p) for p in cluster_top[cluster_top['cluster_label']==cl]['product_id']
            if p not in purchased.get(cid,set())][:3]
    while len(recs)<3: recs.append(None)
    rows.append([int(cid), cl]+recs)

pd.DataFrame(rows,columns=['customer_id','cluster_label',
    'recommended_product_1','recommended_product_2','recommended_product_3']
).to_csv('Session5_Segmentation_and_Recommendations_short.csv', index=False)
print('✅ Saved Session5_Segmentation_and_Recommendations_short.csv')
# จุดสังเกต: cluster_label มีแค่ 1,2,3 | ไฟล์มี 5 คอลัมน์ | recommended ≠ สินค้าที่ซื้อแล้ว
```
