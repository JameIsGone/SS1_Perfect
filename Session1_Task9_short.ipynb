# Session1_Task9_short — Customer Lifetime Value (CLTV)

```python
import pandas as pd, warnings; warnings.filterwarnings('ignore')

s = pd.read_csv('sales_transactions_cleaned.csv')
s['date']    = pd.to_datetime(s['date'])
s['revenue'] = (s['quantity']*s['price']) - pd.to_numeric(s['discount_amount'],errors='coerce').fillna(0)

total_months = s['date'].dt.to_period('M').nunique()  # จำนวนเดือนทั้งหมดในข้อมูล

# CLTV = avg_purchase_value × (transactions/months) × 36
cltv = (s.groupby('customer_id').agg(total_rev=('revenue','sum'), total_tx=('transaction_id','nunique'))
         .assign(cltv=lambda x: (x['total_rev']/x['total_tx'] * x['total_tx']/total_months * 36).round(2))
        [['cltv']].reset_index()
         .astype({'customer_id':int}))

cltv.to_csv('Session1_CLTV_short_short.csv', index=False)
print('✅ Saved Session1_CLTV_short.csv')
display(cltv.head())
# จุดสังเกต: 2 คอลัมน์ (customer_id int, cltv float), cltv > 0 ทุกแถว
```
