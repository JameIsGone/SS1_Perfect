# Session1_Task5_short — Customer Analysis

```python
import pandas as pd, matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages

c = pd.read_csv('customers_cleaned.csv')
c['membership_status'] = c['membership_status'].str.title()  # normalize: basic→Basic

# age groups: 18-24, 25-34, 35-44, 45+
c['age_group'] = pd.cut(c['age'], bins=[17,24,34,44,200], labels=['18-24','25-34','35-44','45+'])
age_counts = c['age_group'].value_counts().sort_index()

# gender % (M และ F เท่านั้น)
g = c[c['gender'].isin(['M','F'])]['gender'].value_counts()
gender_df = pd.DataFrame({'Gender':g.index, 'Percentage (%)':(g/g.sum()*100).round(2).values})

# avg spending per tier
avg = (c[c['membership_status'].isin(['Basic','Silver','Gold'])]
       .groupby('membership_status')['total_spending'].mean().round(2)
       .reindex(['Basic','Silver','Gold']))
tier_df = pd.DataFrame({'Tier':avg.index, 'Avg Spending ($)':avg.values})

def style_tbl(tbl, color):
    tbl.auto_set_font_size(False); tbl.set_fontsize(11); tbl.scale(1.5, 2.2)
    for (r,_),cell in tbl.get_celld().items():
        if r==0: cell.set_facecolor(color); cell.set_text_props(color='white',fontweight='bold')
        cell.set_edgecolor('lightgray')

with PdfPages('Session1_CustomerAnalysis_short.pdf') as pdf:
    # bar chart
    fig,ax = plt.subplots(figsize=(8,5))
    age_counts.plot(kind='bar',color='steelblue',ax=ax,rot=0)
    ax.set(title='Distribution of Customer Age Groups',ylabel='Number of Customers',xlabel='Age Group')
    ax.grid(axis='y',linestyle='--',alpha=0.5); fig.tight_layout()
    pdf.savefig(fig,bbox_inches='tight'); plt.close()
    # gender table
    fig,ax = plt.subplots(figsize=(6,2)); ax.axis('off')
    ax.set_title('Gender Distribution (%)',fontsize=13,fontweight='bold',pad=16)
    style_tbl(ax.table(cellText=gender_df.values,colLabels=gender_df.columns,loc='center',cellLoc='center'),'steelblue')
    fig.tight_layout(); pdf.savefig(fig,bbox_inches='tight'); plt.close()
    # tier table
    fig,ax = plt.subplots(figsize=(6,2)); ax.axis('off')
    ax.set_title('Average Spending per Loyalty Tier',fontsize=13,fontweight='bold',pad=16)
    style_tbl(ax.table(cellText=tier_df.values,colLabels=tier_df.columns,loc='center',cellLoc='center'),'seagreen')
    fig.tight_layout(); pdf.savefig(fig,bbox_inches='tight'); plt.close()

print('✅ Saved Session1_CustomerAnalysis_short.pdf')
# จุดสังเกต: gender % รวม=100, tier ครบ Basic/Silver/Gold, age group ครบ 4 กลุ่ม
```
