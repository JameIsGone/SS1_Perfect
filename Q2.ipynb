```python
import pandas as pd, numpy as np, warnings; warnings.filterwarnings('ignore')

sales = pd.read_csv('sales_transactions.csv')
customers = pd.read_csv('customers.csv')

customers['age'] = customers['age'].fillna(customers['age'].median())
customers['phone_number'] = (customers['phone_number'].fillna('0').astype(str).str.replace(r'[^0-9+]','',regex=True).replace(['','00'],'0'))

sales['promotion_id'] = sales['promotion_id'].fillna(0).astype(int)

fix_dates = lambda col: (
    pd.to_datetime(col, errors='coerce', format='mixed').pipe(lambda x: x.where(x.dt.year.between(2000, 2025))).fillna(pd.Timestamp('2025-01-01')).dt.normalize() 
    + pd.to_timedelta(np.random.randint(9*3600,16*3600+1,size=len(col)), unit='s')
)

customers['join_date'], customers['last_purchase_date'], sales['date'] = fix_dates(customers['join_date']), fix_dates(customers['last_purchase_date']), fix_dates(sales['date'])

customers.to_csv('customers_cleaned.csv',index=False)
sales.to_csv('sales_transactions_cleaned.csv',index=False)
print('Success')
```
